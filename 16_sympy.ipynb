# SymPy

- SymPy is a Python library for symbolic mathematics.

- Symbolic computation deals with the computation of mathematical objects symbolically.
This means that the mathematical objects are represented exactly, not approximately, and mathematical expressions with unevaluated variables are left in symbolic form.

- Admitedly, the Mathematica software might be more powerful.
However, SymPy is a great open-source alternative that is sufficient for many tasks.
Moreover, SymPy can be easily integrated with other Python libraries, making it a versatile tool for scientific computing.
For example, you can prepare a complex symbolic expression in SymPy and then use NumPy to perform fast numerical evaluations or Matplotlib to visualize the results.

# SymPy symbolic variables and expressions

- The symbolic computation keeps mathematical expressions in the precise symbolic form, while numerical computation approximates mathematical expressions using floating-point numbers.

```python
import sympy
import math
print(math.sqrt(8))
print(sympy.sqrt(8))
```

- Expressions in SymPy are built using symbolic variables, which can be defined using the `symbols` function. In its string argument, the variables are separated by spaces.

- N.B. Symbolic variables are treated as complex numbers by default.

```python
import sympy
x, y = sympy.symbols('x y')
expr = x + 2*y
print(expr)
```

- There is a difference between the SymPy symbolic variable, and the regular Python variable used to reference it. The following example illustrates this difference:

```python
import sympy
x, y = sympy.symbols('y x')
expr = x + 2*y
print(expr)
```

- The above code illustrates that the SymPy symbolic variables can have different names than the Python variables used to reference them. However, it is a common practice to use the same names for both to avoid confusion.

- Longer names can also be used for symbolic variables:

```python
import sympy
mu, nu, theta = sympy.symbols('mu nu theta')
expr = (mu + nu) ** (2 - theta)
print(expr)
```

- Expressions are immutable, meaning that any operation on an expression creates a new expression without modifying the original one. For example, adding two expressions together results in a new expression, leaving the original expressions unchanged.

```python
import sympy
x, y = sympy.symbols('x y')
expr1 = x + y
expr2 = x - y
expr3 = expr1 + expr2
print('expr1: ', expr1)
print('expr2: ', expr2)
print('expr3: ', expr3)
```

- Strings can be converted to SymPy expressions using the `sympify` function. This is useful when reading expressions from external sources, such as files or user input.

```python
import sympy
expr = sympy.sympify('x**2 + 2*x + 1')
print(expr)
```

- Pretty printing of expressions is achived by calling the `sympy.init_printing()` function.

```python
import sympy
sympy.init_printing()
x, y = sympy.symbols('x y')
expr = x + y ** 2
expr # Do not use print() here to see pretty output
```

# Numerical evaluation of SymPy expressions

- SymPy expressions can be evaluated numerically using the `evalf` method.

- The `evalf` method computes a numerical approximation of the expression to a specified number of decimal places, given by the `n` parameter.

- To evaluate an expression at specific values of its variables, use the `subs` parameter.

```python
import sympy
x, y = sympy.symbols('x y')
expr = x + y
num_expr = expr.evalf(subs = {x : 1.5, y : 2.5}, n = 5)
print(num_expr)
```

```python
# Calculate pi to 100 decimal places
import sympy
print(sympy.pi.evalf(n = 100))
```

# lambdify function for fast numerical evaluation

- The `evalf` function is good if you want to do simple evaluation in one point, but if you intend to evaluate an expression at many points, there is a more efficient way. For example, if you wanted to evaluate an expression at a thousand points, using SymPy would be far slower than it needs to be, especially if you only care about machine precision. Instead, you should use libraries like NumPy.

- The easiest way to convert a SymPy expression to an expression that can be numerically evaluated is to use the `lambdify()` function. `lambdify()` acts like a lambda function, except it converts the SymPy names to the names of the given numerical library, e.g. NumPy.


```python
import sympy
import numpy
x = sympy.symbols('x')
expr = x ** 2 + 1
f = sympy.lambdify(x, expr, "numpy")
f(numpy.array([0, 1, 2, 3, 4, 5]))
```

# Expansion of expressions

- The `expand()` function expands products and powers.

- There are also functions to perform specific types of expansion, e.g., `expand_trig()` for trigonometric expressions, `expand_log()` for logarithmic expressions, etc.

- See [https://docs.sympy.org/latest/tutorials/intro-tutorial/simplification.html](https://docs.sympy.org/latest/tutorials/intro-tutorial/simplification.html) for more details.

```python
import sympy
x = sympy.symbols('x')
expr = (x + 1) ** 2
expanded_expr = sympy.expand(expr)
expanded_expr
```

```python
# expand_trig() example
import sympy
x = sympy.symbols('x')
expr = sympy.sin(x + sympy.pi / 4)
expanded_expr = sympy.expand_trig(expr)
expanded_expr
```

- If an identity is not valid for all complex numbers, then the expansion does not happen. For example, the identity $\log(x y) = \log(x) + \log(y)$ not valid due to the $2\pi$ discontinuity of the logarithm (e.g. $\log(-i) = -i\pi/2 \neq \log(-1) + \log(i)$). However, it is always valid for real positive values of $x$ and $y$. Therefore, the expansion only happens if we specify that the symbolic variables are positive.

```python
import sympy
x, y = sympy.symbols('x y')                    # the logarithm expansion will not happen
# x, y = sympy.symbols('x y', positive = True) # the logarithm expansion will happen
expr = sympy.log(x * y)
expanded_expr = sympy.expand_log(expr)
expanded_expr
```

# Simplification of expressions

- The `simplify()` function attempts to simplify expressions to a more compact form.

- There are functions to perform specific types of simplification, e.g., `trigsimp()` for trigonometric expressions, `logcombine()` for logarithmic expressions, `factor()` for factoring polynomials, etc.

- N.B. `simplify()` can be unnecessarily slow, since it tries many kinds of simplifications before picking the best one. If you already know exactly what kind of simplification you are after, it is better to apply the specific simplification function(s) that apply those simplifications.

- See [https://docs.sympy.org/latest/tutorials/intro-tutorial/simplification.html](https://docs.sympy.org/latest/tutorials/intro-tutorial/simplification.html) for more details.

```python
import sympy
x = sympy.symbols('x')
expr = sympy.sin(x) ** 2 + sympy.cos(x) ** 2
simpl_expr = sympy.simplify(expr)
simpl_expr
```

```python
# trigsimp() example
import sympy
x = sympy.symbols('x')
expr = sympy.sin(x) ** 2 - sympy.cos(x) ** 2
simpl_expr = sympy.trigsimp(expr)
simpl_expr
```

```python
# logcombine() example
import sympy
x = sympy.symbols('x', positive = True) # specify that x is positive to avoid issues with logarithm domain; otherwise, simplification may not proceed
expr = sympy.log(x) + sympy.log(x + 1)
simpl_expr = sympy.logcombine(expr)
simpl_expr
```

```python
# factor() example
import sympy
x = sympy.symbols('x')
expr = x**2 - 1
factored_expr = sympy.factor(expr)
factored_expr
```

# Derivatives

- The `diff()` function computes derivatives of expressions with respect to specified variables.

- There are the `diff()` function and the expressions' `diff()` method, which can be used interchangeably.

```python
# diff() function example
import sympy
x, y = sympy.symbols('x y')
expr = x**2 * y + sympy.sin(x * y)
deriv_expr = sympy.diff(expr, x)
deriv_expr
```

- The same functionality is also provided by the `diff()` method of expressions.

```python
# diff() method example
import sympy
x, y = sympy.symbols('x y')
expr = x**2 * y + sympy.sin(x * y)
deriv_expr = expr.diff(x)
deriv_expr
```

- Several derivatives can be computed at once. In the following example, we compute the third derivative with respect to `x` and the second derivative with respect to `y`.

```python
import sympy
x, y = sympy.symbols('x y')
expr = x**2 * y + sympy.sin(x * y)
deriv_expr = expr.diff(x, x, x, y, y)
deriv_expr
```

- Derivatives of unspecified order can be created using tuple `(x, n)` where `n` is the order of the derivative with respect to `x`.

```python
import sympy
x, n = sympy.symbols('x n')
expr = x**2 * sympy.sin(x)
deriv_expr = expr.diff((x, n))
deriv_expr
```

- It is possible to treat the derivative operator as a symbolic object using the `Derivative` class. This allows for manipulation of derivatives without immediately evaluating them.

```python
import sympy
x, y = sympy.symbols('x y')
expr = x**2 * y + sympy.sin(x * y)
deriv_expr = sympy.Derivative(expr, x, x, x)
deriv_expr
```

- The delayed derivative can be evaluated using the `doit()` method.

```python
deriv_expr = deriv_expr.doit()
deriv_expr
```

# Integrals

- To compute an integral, use the `integrate()` function.

- There are two kinds of integrals, definite and indefinite.

- Indefinite integral: just pass the variable after the expression in the `integrate()` function.

```python
import sympy
x = sympy.symbols('x')
expr = x
int_expr = sympy.integrate(expr, x)
int_expr
```

- Note that the integration constant is omitted by SymPy.

- The same functionality is also provided by the `integrate()` method of expressions.

```python
import sympy
x = sympy.symbols('x')
expr = x
int_expr = expr.integrate(x)
int_expr
```

- Definite integral: pass the variable and the limits as a tuple to the `integrate()` function.

```python
import sympy
x = sympy.symbols('x')
expr = x
int_expr = expr.integrate((x, 0, 1))
int_expr
```

- The symbol for infinity is `sympy.oo`. This is because oo looks like $\infty$, and is easy to type.

```python
import sympy
x = sympy.symbols('x')
expr = sympy.exp(-x)
int_expr = expr.integrate((x, 0, sympy.oo))
int_expr
```

- As with `Derivative`, you can create an unevaluated integral using `Integral`. To later evaluate this integral, call `doit()`.

```python
import sympy
x = sympy.symbols('x')
expr = sympy.exp(-x)
int_expr = sympy.Integral(expr, x)
int_expr
```

```python
int_expr.doit()
```

# Numerical integration

- SymPy also provides support for numeric integration. It leverages the precision capabilities of the mpmath library to enhance the accuracy of numeric integration calculations.

- Use the `evalf()` method on an `Integral` object to perform numerical integration. You can define the number of significant digits using the `n` parameter.

```python
import sympy
x = sympy.symbols('x')
expr = sympy.exp(-x)
int_expr = sympy.Integral(expr, (x, 0, 1))
int_expr.evalf(30)
```

- You can always lambdify the integrand and use numerical integration methods from libraries like SciPy for potentially better performance on complex integrals.

```python
import sympy
import scipy.integrate
x = sympy.symbols('x')
expr = sympy.exp(-x)
f = sympy.lambdify(x, expr, 'scipy')
print(scipy.integrate.quad(f, 0, 1))
```

- Note that `scipy.integrate.quad` returns a tuple containing the integral value and an estimate of the absolute error.

# Limits

- SymPy can compute symbolic limits with the limit() function. The syntax to compute $$\lim_{x \to a} f(x)$$ is `sympy.limit(expr, x, a)`.

```python
import sympy
x = sympy.symbols('x')
expr = sympy.sin(x) / x
limit_expr = sympy.limit(expr, x, 0)
limit_expr
```

- To evaluate a limit at one side only, pass '+' or '-' as a fourth argument to limit(). For example, to compute $$\lim_{x \to 0^+} 1/x$$ use `sympy.limit(1/x, x, 0, '+')`.

```python
import sympy
x = sympy.symbols('x')
expr = 1 / x
limit_expr = sympy.limit(expr, x, 0, '+')
limit_expr
```

- Similarly, for limits from the other side:

```python
import sympy
x = sympy.symbols('x')
expr = 1 / x
limit_expr = sympy.limit(expr, x, 0, '-')
limit_expr
```

# Equations

- SymPy solves equations using the `solveset()`, `linsolve()`, `nonlinsolve()`, `root()`, `dsolve()`, and other functions.

- For more details, and limitations, see [https://docs.sympy.org/latest/tutorials/intro-tutorial/solvers.html](https://docs.sympy.org/latest/tutorials/intro-tutorial/solvers.html).

- Equation objects can be created using the `Eq()` function. Then, it is passed to the solving function together with the variable(s) to solve for.

```python
import sympy
x = sympy.symbols('x')
eq = sympy.Eq(x**2, 1) # Syntax: Eq(left_side, right_side)
sympy.solveset(eq, x)
```

- An expression can also be passed directly to the solving function, in which case it is assumed to be equal to zero. The following is equivalent to the previous example:

```python
import sympy
x = sympy.symbols('x')
expr = x**2 - 1  # Expression assumed to be equal to zero
sympy.solveset(expr, x)
```

# Systems of equations

- SymPy can solve systems of linear equations using the `linsolve()` function.

```python
import sympy
x, y, z = sympy.symbols('x y z')
system = [x + y + z - 1,
          x + y + 2*z - 3 ]
sympy.linsolve(system, (x, y, z))
```

- Nonlinear systems of equations are solved using `nonlinsolve()`

```python
import sympy
x, y = sympy.symbols('x y')
system = [x**2 - 2*y**2 -2,
          x*y - 2]
sympy.nonlinsolve(system, (x, y))
```

# Differential equations

- SymPy can solve ordinary differential equations (ODEs) using the `dsolve()` function.

- First, create an undefined function by passing `cls = sympy.Function` to the `symbols()` function.

```python
import sympy
f, g = sympy.symbols('f g', cls = sympy.Function)
```

- Then, prepare the differential equation using the `diff()` method of the undefined function, and pass it to the `dsolve()` function along with the function to solve for.

```python
diffeq = sympy.Eq(f(x).diff(x, x) - 2 * f(x).diff(x) + f(x), sympy.sin(x))
sympy.dsolve(diffeq, f(x))
```

# Probability distribution functions

- Each measurement of whatever complexity is imagined like drawing one or several random numbers from a PDF.

- Even if you take a huge dataset, but measure just one quantity, you are still drawing one random number from a PDF.

- Imagine you measure the mass of some object. You use your balance 1000 times, and you get 1000 different values. Evaluating the mean is equivalent to drawing one random number from some PDF. The mean is a random variable because it depends on the 1000 random numbers you got from the balance.

- Let's model the balance as a Normal distribution with $\mu = 100$ g and $\sigma = 5$ g:

```python
# Simple model of weighting with a balance: Normal distribution with mean 100 g and standard deviation 5 g.
from scipy.stats import norm
import numpy as np
import matplotlib.pyplot as plt
mu = 100
sigma = 5

x = np.linspace(80, 120, 2000)
balance_pdf = norm.pdf(x, loc = mu, scale = sigma)
plt.plot(x, balance_pdf)
plt.xlabel(r'$m$ [g]')
plt.ylabel('Probability density')
plt.title('PDF of the balance')
plt.show()
```

- Let's weight the object 1000 times, which is equivalent to drawing 1000 random numbers from the PDF of the balance. Let's plot each value as a dot in a scatter plot.

```python
# Draw 1000 random numbers from the PDF of the balance.
N = 1000
samples = norm.rvs(loc = mu, scale = sigma, size = N)

# Plot the scatter plot of the 1000 values.
plt.scatter(range(N), samples, alpha = 0.5)
# Indicate the +- 1 sigma interval with horizontal lines.
plt.axhline(y = mu + sigma, color = 'r', linestyle = '--', label = r'$1 \sigma$')
plt.axhline(y = mu - sigma, color = 'r', linestyle = '--')
# Write +- 1 sigma in the plot.
plt.text(0, mu + sigma + 0.5, r'$\mu + 1\sigma$', color = 'r')
plt.text(0, mu - sigma - 1.5, r'$\mu - 1\sigma$', color = 'r')
# Write +- 2 sigma in the plot.
plt.axhline(y = mu + 2 * sigma, color = 'g', linestyle = '--', label = r'$2 \sigma$')
plt.axhline(y = mu - 2 * sigma, color = 'g', linestyle = '--')
plt.text(0, mu + 2 * sigma + 0.5, r'$\mu + 2\sigma$', color = 'g')
plt.text(0, mu - 2 * sigma - 1.5, r'$\mu - 2\sigma$', color = 'g')
plt.xlabel('Sample number')
plt.ylabel(r'$m$ [g]')
plt.title('Scatter plot of 1000 samples from the PDF of the balance')
plt.show()
```

- We estimate the mass as a mean of the $N = 1000$ values $\{m_i\}$.

$$
\hat{m} = \frac{1}{N} \sum_{i=1}^N m_i
$$

- We estimate the uncertainty of the mean as usual:

$$
\hat{\sigma}_{\hat{m}} = \frac{\hat{\sigma}}{\sqrt{N}} = \frac{1}{\sqrt{N}} \sqrt{\frac{1}{N-1} \sum_{i=1}^N (m_i - \hat{m})^2}
$$

- Note that $\hat{\sigma}$ is the estimate of the balance PDF parameter, while $\hat{\sigma}_{\hat{m}}$ is the uncertainty in the mean estimate, $\hat{m}$.

```python
# Estimate the mean and its uncertainty.
mhat = np.mean(samples)
sigmahat_m = np.std(samples, ddof = 1) / np.sqrt(len(samples))
print(f'Estimated mean: mhat = {mhat:.2f} g')
print(f'Estimated uncertainty of the mean: sigmahat_m {sigmahat_m:.2f} g')
```

- The PDF for $\hat{m}$ is a Normal distribution with $\mu' = 100$ g and $\sigma' =  \sigma / \sqrt{N}$.

- Let's plot the PDF of $\hat{m}$ in one plot with the PDF of the balance to see how the width is reduced compared to the balance PDF.

```python
# Plot the PDF of mhat and the balance PDF.
mhat_pdf = norm.pdf(x, loc = mu, scale = sigma / np.sqrt(len(samples)))
plt.plot(x, balance_pdf, label = 'Balance PDF')
plt.plot(x, mhat_pdf, label = r'PDF of $\hat{m}$')
plt.xlabel(r'$m$ [g]')
plt.ylabel('Probability density')
plt.title(r'PDF of the balance and the PDF of $\hat{m}$')
plt.legend()
plt.show()
```

Summary: the whole measurement is really like drawing one random number ($\hat{m}$) from some PDF (the PDF of $\hat{m}$). Thanks to the huge number of weightings, the PDF of $\hat{m}$ is much narrower than the PDF of the balance. Therefore, the uncertainty in $\hat{m}$ is estimated to be much smaller than the parameter $\sigma$ of the balance PDF.

# PDFs in SciPy

- SciPy has a huge number of PDFs implemented in the `scipy.stats` module. You can find the full list [here](https://docs.scipy.org/doc/scipy/reference/stats.html).

- Each PDF is implemented as a class with methods for evaluating the PDF, CDF, and other properties. For example, the Normal distribution is implemented as `scipy.stats.norm`. Let's see how to use it:

```python
from scipy.stats import norm
# Evaluate the PDF of the Normal distribution at three values of the x variable.
x_values = [2, 4, 5]
pdf_values = norm.pdf(x_values, loc = 5, scale = 3)
print(f'PDF values at x = {x_values}: {pdf_values}')
```

- The parameters `loc` and `scale` are important! Most methods of the PDF classes have these parameters. Their exact meaning depends on the PDF. They are defined such that

$$
\mathrm{pdf}(x, \mathrm{loc}, \mathrm{scale})\ \equiv\ \frac{1}{\mathrm{scale}} \mathrm{pdf}\left(\frac{x - \mathrm{loc}}{\mathrm{scale}}\right)
$$

- For the Normal distribution, `loc` is the mean, $\mu$, and `scale` is the width, $\sigma$.

- Cumulative distribution function (CDF) is the integral of the PDF from $-\infty$ to $x$:
$$
\mathrm{cdf}(x, \mathrm{loc}, \mathrm{scale})\ \equiv\ \mathrm{cdf}\left(\frac{x - \mathrm{loc}}{\mathrm{scale}}\right)
$$

```python
# Evaluate the CDF of the Normal distribution at three values of the x variable.
cdf_values = norm.cdf(x_values, loc = 5, scale = 3)
print(f'CDF values at x = {x_values}: {cdf_values}')
```

- To get percentiles, use the percent point function (PPF), which is the inverse of the CDF:

```python
# Evaluate the 16th, 50th, and 84th percentiles of the Normal distribution.
integrals = [0.16, 0.50, 0.84]
percentiles = norm.ppf(integrals, loc = 5, scale = 3)
print(f'Percentiles at {integrals}: {percentiles}')
```

- An important functionality of the PDF classes is the ability to draw random numbers from the PDF. This is done with the `rvs` method, which stands for "random variates". For example, to draw 10 random numbers, we can do:

```python
# Draw 10 random numbers and print them.
random_numbers = norm.rvs(loc = 5, scale = 3, size = 10)
print(random_numbers)
```

# PDFs in NumPy

- NumPy does not provide analytical PDFs, so you cannot, e.g., evaluate the PDF or CDF at a point.

- However, you can draw random numbers from a PDF with the `numpy.random` module. For example:

```python
# Draw 10 random numbers from the Normal distribution.
import numpy as np
random_numbers = np.random.normal(loc = 5, scale = 3, size = 10)
print(random_numbers)
```

- Note that the `loc` and `scale` parameters are the same as in SciPy, so you can use the same values for both libraries.

- NumPy provides methods for the data analysis of the random numbers you draw, such as `numpy.mean` and `numpy.std`, which can be used to estimate the mean and standard deviation of the PDF from which you drew the random numbers.

```python
# Estimate the mean and standard deviation of the random numbers you drew.
estimated_mean = np.mean(random_numbers)
estimated_std = np.std(random_numbers, ddof = 1)
print(f'Estimated mean: {estimated_mean:.2f}')
print(f'Estimated standard deviation: {estimated_std:.2f}')
```

# Multivariate PDFs

- SciPy also provides multivariate PDFs, such as [scipy.stats.multivariate_normal](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.multivariate_normal.html), which is a multivariate Normal distribution. The parameters of the multivariate Normal distribution are the mean vector and the covariance matrix.

- The formula for the multivariate Normal distribution contains the covariance matrix:
$$
\mathrm{pdf}(\mathbf{x}, \boldsymbol{\mu}, \boldsymbol{V})\ \equiv\ \frac{1}{(2\pi)^{k/2} |\boldsymbol{V}|^{1/2}} \exp\left(-\frac{1}{2} (\mathbf{x} - \boldsymbol{\mu})^T \boldsymbol{V}^{-1} (\mathbf{x} - \boldsymbol{\mu})\right)
$$

- The Normal distribution is the only distribution that is fully described by its mean and covariance. Other PDFs need more moments to be fully described (infinite number of them in general).

```python
# Draw 2D normal distribution with mean vector [0, 0] and covariance matrix [[3, 2], [2, 2]]
# as a heatmap.
from scipy.stats import multivariate_normal
import numpy as np
import matplotlib.pyplot as plt
mean = [0, 0]
cov = [[3, 2], [2, 2]]
x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)
X, Y = np.meshgrid(x, y)
pos = np.dstack((X, Y))
Z = multivariate_normal.pdf(pos, mean, cov)
plt.contourf(X, Y, Z, levels = 50, cmap = 'viridis')
plt.colorbar(label = 'Probability density')
plt.xlabel('x')
plt.ylabel('y')
plt.title('PDF of the multivariate Normal distribution')
plt.show()
```

# Estimation of the covariance matrix

- The covariance matrix is a generalization of the variance to multiple dimensions. It describes the spread of the data in multiple dimensions and the correlations between the dimensions.

- When measuring multiple quantities, you must estimate the covariance matrix, as it holds the information about the uncertainties (not just its diagonal elements, but all of them).

- If we draw several pairs of values from a 2D distribution, then the covariance matrix is estimated as follows:
    $$
    \hat{V}_{ij} = \frac{1}{N-1} \sum_{k=1}^N (x_{i,k} - \hat{\mu}_i) (x_{j,k} - \hat{\mu}_j)
    $$
    where $x_{i,k}$ is the $i$-th component of the $k$-th drawn pair, and $\hat{\mu}_i$ is the estimated mean of the $i$-th component.

- Explanation: the formula above is approximating the ideal, but impossible theoretical calculation. In the calculation, points are given a high weight if they are in a region of high probability density:
    $$
    V_{ij} = \int \Pi_k \mathrm{d}x_k (x_i - \mu_i) (x_j - \mu_j) \mathrm{pdf}(x_k)
    $$
    If we draw pairs of values from the PDF, then we get more points in the region of high probability density, so we can approximate the integral with a sum over the drawn pairs, which is the formula above.

- In many cases, we don't have data to estimate the covariance matrix this way, however. E.g. when we measure values of some function parameters by fitting the model to data, we have just one pair (or one N-tuple, in general) of values as the outcome. Fortunately, there are other methods to estimate the covariance matrix, such as the Fisher or Hessian matrix (second derivatives of the likelihood or the loss function). This technique is discussed [elsewhere](https://youtu.be/xwcVRAUpRkU?si=jCMjxxMH9hOHph0m), but it is widely used in physics. Be aware that tools like `scipy.optimize.curve_fit` use it or something similar.

- In some cases, no analytical method works... Fortunately, we have Monte Carlo methods, based on random number generation! See the end of this notebook.

# Using the covariance matrix

- Whenever you evaluate an expression that depends on several random variables, $y = y(x_1, x_2, \ldots, x_n)$, you need to use the covariance matrix to evaluate the uncertainty of the expression. This is done with the error propagation formula:
  $$
  \sigma_{\hat{y}} = \sqrt{\sum\limits_{i,j=1}^n \frac{\partial y}{\partial x_i}\Big|_{\vec x=\hat{\vec{x}}} \frac{\partial y}{\partial x_j}\Big|_{\vec x=\hat{\vec{x}}} V_{ij}}
  $$
  In case of just two variables, the formula is:
    $$
    \sigma_{\hat{y}} = \sqrt{\left(\frac{\partial y}{\partial x_1}\Big|_{\vec x=\hat{\vec{x}}}\right)^2 V_{11} + \left(\frac{\partial y}{\partial x_2}\Big|_{\vec x=\hat{\vec{x}}}\right)^2 V_{22} + 2 \frac{\partial y}{\partial x_1}\Big|_{\vec x=\hat{\vec{x}}} \frac{\partial y}{\partial x_2}\Big|_{\vec x=\hat{\vec{x}}} V_{12}}
    $$

- In pactice, use tools implementing the error propagation formula, such as `uncertainties` in Python. They will take care of the derivatives and everything for you. As an example, let's assume we measured the length and width of a rectangle, and we want to calculate the area and its uncertainty (let's not talk about where the correlation between the length and width comes from):

```python
import uncertainties

nominal_values    = [1, 2]
covariance_matrix = [[0.01, -8.e-3 ],
                     [-8.e-3 , 0.03]]
(a, b) = uncertainties.correlated_values(nominal_values, covariance_matrix)
print(a * b)

# Let's calculate the same thing, ignoring the correlation between the two variables.
a_uncorr = uncertainties.ufloat(a.nominal_value, a.std_dev)
b_uncorr = uncertainties.ufloat(b.nominal_value, b.std_dev)
print(a_uncorr * b_uncorr)
```

# Pseudo-experiments (Monte Carlo method)

- In some cases, no analytical method works to estimate the covariance matrix. In such cases, we can use the Monte Carlo method, which is based on random number generation.

- Pros: 

  - Very little assumptions are needed.

  - Works in any case, as long as you can draw random numbers from the PDF of the measured quantities and evaluate the expression you are interested in.

- Cons:

  - Computationally expensive, as you need to draw a large number of random numbers and evaluate the expression for each of them.

- Best to explain with an example. Let's say we measured the length of a square, $\hat{a}$, and its uncertainty, $\hat{\sigma}_a$, and we want to calculate the area of the square and its uncertainty.

  - We will generate a set of pseudo-measurements of the length, $\{\hat{a}_i\}_{i=1}^N$, by drawing random numbers from the PDF of $\hat{a}$.

  - The PDF of $\hat{a}$ is a Normal distribution with mean $a$ and standard deviation $\sigma_a$. As we don't know the true values of $a$ and $\sigma_a$, we will use the estimates $\hat{a}$ and $\hat{\sigma}_a$ as the parameters of the PDF.

  - For each pseudo-measurement $\hat{a}_i$, we will calculate the area of the square, $\hat{A}_i = \hat{a}_i^2$.
  
  - We will estimate the variance of the PDF of the area as the variance of the set $\{\hat{A}_i\}_{i=1}^N$
  $$
  \hat{\sigma}_A^2 = \frac{1}{N} \sum_{i=1}^N (\hat{A}_i - \hat{A})^2
  $$
  where $\hat{A} = \hat{a}^2$ is our area estimate based on the measured length, $\hat{a}$.

```python
# Code for the above example.
import numpy as np

# Set some numerical values for the measured length and its uncertainty.
ahat = 5
sigmahat_a = 2
N = 1000

# Generate N pseudo-experiments.
a_pseudo = np.random.normal(ahat, sigmahat_a, N)

# Calculate the areas for each pseudo-experiment.
A_pseudo = a_pseudo ** 2

# Estimate the mean and standard deviation of the area.
Ahat = np.mean(A_pseudo)
sigmahat_A = np.std(A_pseudo, ddof = 0)

print(f'Estimated uncertainty in area: {sigmahat_A:.2f} m^2')

# Compare with the error propagation formula.
import uncertainties
a = uncertainties.ufloat(ahat, sigmahat_a)
A = a ** 2
print(f'Estimated uncertainty in area with error propagation formula: {A.std_dev:.2f} m^2')
```

One more example: Estimating the covariance matrix in the "Beta-radiation absorption" homework using pseudo-experiments.

- Let's start with a solution of the homework:

```python
import matplotlib.pyplot as plt
import uncertainties
import numpy as np
import pandas as pd
from scipy.optimize import curve_fit

data_str = """
d      N
0.0000 13556
0.0097 10336
0.0195  8466
0.0293  6906
0.0390  6176
0.0487  5256
0.0586  4865
0.0689  4264
0.0791  3872
0.0896  3562
0.1002  3257
0.1112  3071
0.1225  2812
0.1410  2321
0.2810  1037
0.4150   496
0.5520   240
0.6931   159
0.8301   117
0.9616   119
1.0954   101
"""

# Read your measured data from the string `data_str` into a DataFrame whose columns are named
#  - 'd' for the absorber thickness,
#  - 'N' for the number of events.
# Add a column 'N_unc' to the DataFrame that contains the uncertainty on N.

from io import StringIO

# Create the pandas dataframe from the string above.
df = pd.read_table(StringIO(data_str), sep = "\s+")

# Define the absorption function
def absorption_curve(d, N_1, mu_E1, N_2, mu_E2, N_B):
    return N_1 * np.exp(-mu_E1 * d) + N_2 * np.exp(-mu_E2 * d) + N_B

# For a given thickness array d, and counts array N, estimate the absorption_curve parameter values.
def estimate_parameters(d, N):
    nom, cov = curve_fit(absorption_curve, d, N, sigma = N ** 0.5, absolute_sigma = True)
    return nom

# Calculate the nominal estimates of the parameters.
nom_estimates = estimate_parameters(df['d'], df['N'])

pseudoexperiments = []

# Generate 100 pseudo-experiments by generating the counts N for each thickness d according to a Poisson distribution with mean given by df['N'].
n_toys = 1000
n_failed_fits = 0
for i in range(n_toys):
    if i % 10 == 0:
        print(f'Generating pseudo-experiment {i} / {n_toys}')
    N_pseudo = np.random.poisson(df['N'])
    try:
        pseudo = estimate_parameters(df['d'], N_pseudo)
        # swap the order of the parameters to match the order in nom_estimates
        if abs(pseudo[0] - nom_estimates[0]) > abs(pseudo[0] - nom_estimates[2]):
            pseudo = [pseudo[2], pseudo[3], pseudo[0], pseudo[1], pseudo[4]]
        pseudoexperiments.append(pseudo)
    except:
        print(f'Failed to fit pseudo-experiment {i}')
        n_failed_fits += 1
        
# pseudoexperiments is a list of 100 arrays, each array containing the estimated parameters for one pseudo-experiment. We can convert it to a numpy array of shape (100, 5).
pseudoexperiments = np.array(pseudoexperiments)
```

```python

# Calculate the covariance matrix of the estimated parameters across the pseudo-experiments.
cov_matrix = np.cov(pseudoexperiments, rowvar = False)

# Estimate the covariance matrix with the curve_fit function and compare it with the covariance matrix estimated with the pseudo-experiments.
nom, cov = curve_fit(absorption_curve, df['d'], df['N'], sigma = df['N'] ** 0.5, absolute_sigma = True)

# Element-by-element comparison of the two covariance matrix elements, and their relative differences.
param_names = ['N_1', 'mu_E1', 'N_2', 'mu_E2', 'N_B']
idx = pd.MultiIndex.from_product([param_names, param_names], names=['row', 'col'])

comparison_df = pd.DataFrame(
    {
        'cov_matrix (toys)': cov_matrix.ravel(),
        'cov (curve_fit)': cov.ravel(),
    },
    index=idx
)

# Calculate the relative difference between the two covariance matrix elements,
# and add it as a third column to the DataFrame.
den = comparison_df['cov (curve_fit)'].to_numpy()
num = (comparison_df['cov_matrix (toys)'] - comparison_df['cov (curve_fit)']).to_numpy()
comparison_df['rel. diff.'] = np.where(np.isclose(den, 0), np.nan, num / den)

# Show the table.
display(
    comparison_df.style
    .format({
        'cov_matrix (toys)': '{:.3e}',
        'cov (curve_fit)': '{:.3e}',
        'rel. diff.': '{:+.2%}'
    })
    .set_caption('Covariance comparison: pseudo-experiments vs curve_fit')
)
```

```python
# Implement the formula for the covariance matrix estimate by hand, without using np.cov.
mean_params = np.mean(pseudoexperiments, axis = 0)
cov_matrix_manual = np.zeros((5, 5))
for i in range(n_toys - n_failed_fits):
    diff = pseudoexperiments[i] - mean_params
    cov_matrix_manual += np.outer(diff, diff)
cov_matrix_manual /= (n_toys - n_failed_fits - 1)
```

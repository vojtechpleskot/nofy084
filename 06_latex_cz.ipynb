# Balíčky

- Balíčky LaTeXu jsou rozšíření, která poskytují dodatečnou funkcionalitu. Pomocí balíčků můžete do dokumentu přidat nové příkazy, prostředí a možnosti formátování.

- Chcete-li použít balíček, musíte ho zahrnout do preambule LaTeXového dokumentu pomocí příkazu `\usepackage{}`.

- Například pro použití balíčku `hyperref`, který umožňuje vytvářet hypertextové odkazy v dokumentu, přidejte do preambule následující řádek:

    ```latex
    \usepackage{hyperref}
    ```

# Literatura

- LaTeX má výkonný systém správy literatury zvaný BibLaTeX. Umožňuje vytvořit databázi bibliografických záznamů a citovat reference v dokumentu.

- Dobrý úvod do BibLaTeXu: [https://cs.overleaf.com/learn/latex/Bibliography_management_with_biblatex](https://cs.overleaf.com/learn/latex/Bibliography_management_with_biblatex)

- Pět hlavních kroků pro použití BibLaTeXu:

    1. Použijte balíček `biblatex` ve svém LaTeXovém dokumentu přidáním příkazu, jako je následující, do preambule:

        ```latex
        \usepackage[
          backend=biber,
          style=numeric,
          sorting=nyt,
          giveninits=true
        ]{biblatex}
        ```
    
    2. Vytvořte soubor s bibliografickou databází (např. `references.bib`) a přidejte do něj své reference ve formátu BibLaTeX.

    3. Zahrňte bibliografii do LaTeXového dokumentu pomocí příkazu `\addbibresource{references.bib}` v preambuli.

    4. Citujte reference v dokumentu pomocí příkazu `\cite{key}`, kde `key` je identifikátor reference v bibliografickém souboru.

    5. Vytiskněte bibliografii před přílohou (pokud existuje) pomocí příkazu `\printbibliography`.

- Hezký BibLaTeXový záznam vypadá takto:

    ```bibtex
    @Article{higgs-discovery-atlas,
        author         = "{ATLAS Collaboration}",
        title          = "{Observation of a new particle in the search for the Standard Model Higgs boson with the ATLAS detector at the LHC}",
        journal        = "Phys. Lett. B",
        volume         = "716",
        year           = "2012",
        pages          = "1",
        doi            = "10.1016/j.physletb.2012.08.020",
        reportNumber   = "CERN-PH-EP-2012-218",
        eprint         = "1207.7214",
        archivePrefix  = "arXiv",
        primaryClass   = "hep-ex",
    }
    ```

    - Ne všechna pole jsou v každém případě použití povinná.

    - V textu můžete citovat tuto referenci pomocí `\cite{higgs-discovery-atlas}`.

- Ke správě literatury a exportu ve formátu BibLaTeX můžete použít online nástroje jako [Zotero](https://www.zotero.org/). Velmi často však stačí kliknout na odkaz ke stažení citace na stránce požadovaného vědeckého článku a zkopírovat záznam BibLaTeX nebo BibTeX (oba mají stejný formát) do souboru `.bib`.

# Vkládání obsahu z jiných souborů .tex

- Použijte příkaz `\input{}`. Pomůže vám to organizovat dokument do více souborů.

- Například pokud máte soubor `introduction.tex` obsahující úvodní část dokumentu, můžete ho zahrnout do hlavního LaTeXového souboru pomocí následujícího příkazu:

    ```latex
    \input{introduction.tex}
    ```

    - DŮLEŽITÉ: Pokud by byl `introduction.tex` v podsložce, např. s názvem `source`, napsali bychom `\input{source/introduction.tex}`.

- Můžete například uchovávat tabulky v samostatných souborech a vkládat je na příslušné místo. To je obzvláště užitečné, pokud tabulky generujete automaticky z dat pomocí skriptu. Skript může vygenerovat soubor `.tex` s tabulkou, který pak vložíte do libovolného souboru `.tex` pomocí `\input{}`.

#### Stringy v Pythonu, stručné shrnutí

- f-stringy jsou nezbytné pro vkládání dat (uložených v proměnných) do stringů.

    - Vložte hodnotu proměnné do stringu pomocí syntaxe se složenými závorkami `{}`. Například:

````python
name, age = "Alice", 30
row = f"{{{name}}} & {age} \\\\"
print(row)
````

- Chcete-li do stringu vložit znak pro zpětné lomítko `\`, musíte ho escapovat dalším zpětným lomítkem (tj. `\\`).

    - Můžete také použít raw string (např. `r"Path\to\file"`) a vyhnout se tak nutnosti escapovat zpětná lomítka.

- Chcete-li do f-stringu vložit znak pro složenou závorku `{` nebo `}`, musíte ji zdvojit (tj. `{{` nebo `}}`).

- f-stringy umožňují výkonné zaokrouhlování!

````python
value = 3.141592653589793
print(f"Value rounded to 2 decimal places: {value:.2f}")
print(f"Value rounded to 2 significant figures: {value:.2g}")
````

- Zápis LaTeXové tabulky do souboru `.tex`:

````python
name, age = "Alice", 30
my_table = \
f"""
\\begin{{tabular}}{{|c|c|}}
\\hline
Name & Age \\\\
\\hline
{name} & {age} \\\\
\\hline
\\end{{tabular}}
"""
with open('table_1.tex', 'w', encoding = "utf-8") as f:
    f.write(my_table)
````

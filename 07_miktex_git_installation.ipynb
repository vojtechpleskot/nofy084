# MiKTeX

- Convenient LaTeX distribution for Windows.

- It's an alternative to the TeX Live distribution.

- MiKTeX doesn't require you to install the full (more than 5 GB) TeX installation. Instead, you install just the necessary 150 MB or so, and you install additional packages, as the need for them comes. During MiKTeX installation, you can choose whether the additional packages are to be installed automatically, when needed, or whether MiKTeX should ask you for permission.

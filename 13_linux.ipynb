# Directory Navigation

Directories are organized in a hierarchical structure.
The top-level directory is called the root directory, denoted simply by `/`. Below the root directory, there are various subdirectories that contain files and other directories.
Each system user has a home directory, typically located at `/home/username`, e.g. `/home/plesv6am`, where they can store their personal files and directories.

To navigate through the directory structure, you can use the `cd` command. For example, to change to my home directory on Chimera, you can use:

````python
cd /home/pleskot
````

To list the contents of the current directory, you can use the `ls` command. This will display the files and directories in the current location.

````python
ls
````

Note that the directory names have a different color than the file names, which makes it easier to distinguish between them. 

To create a new directory, you can use the `mkdir` command followed by the name of the directory you want to create. For example, `mkdir my_directory` will create a new directory called `my_directory` in the current working directory.

````python
mkdir my_directory
````

````python
cd my_directory
````

The `..` symbol represents the parent directory, which is the directory that contains the current directory. For example, if you are in the `/home/plesv6am/my_directory` directory and you want to navigate to its parent directory, you can use the command `cd ..`, which will take you to `/home/plesv6am`. You can also use `cd ../..` to go up two levels in the directory hierarchy, and so on.

````python
cd ..
````

You can also list the contents of a different directory, combining `..` with directory names in the path.

````python
ls ../../etc
````

# Absolute and Relative Paths

When navigating the file system, you can use either absolute or relative paths to specify the location of files and directories.

- An absolute path is a complete path that starts from the root directory and specifies the location of a file or directory. For example, `/home/plesv6am/my_directory` is an absolute path that points to the `my_directory` located in the home directory of the user `plesv6am`.

- A relative path, on the other hand, specifies the location of a file or directory relative to the current working directory. For example, if you are currently in the `/home/plesv6am` directory and you want to access the `my_directory`, you can simply use `my_directory` as a relative path. If you want to access a file called `file.txt` inside `my_directory`, you can use `my_directory/file.txt` as a relative path.


# Permissions

In Linux, each file and directory has associated permissions that determine who can read, write, or execute them. Permissions are represented by a combination of letters and symbols. The three main types of permissions are:

- Read (r): Allows the user to read the contents of a file or list the contents of a directory.

- Write (w): Allows the user to modify the contents of a file or create, delete, or rename files within a directory.

- Execute (x): Allows the user to execute a file as a program or script.

Permissions are typically displayed in a format like `rwxr-xr-x`, where the first three characters represent the owner's permissions, the next three represent the group's permissions, and the last three represent the permissions for others. In this example, the owner has read, write, and execute permissions, while the group and others have read and execute permissions only.

Users of a linux system are organized into groups, and each file or directory is owned by a specific user and group. The owner of a file or directory has the ability to change its permissions using the `chmod` command, while the owner can also change the ownership of a file or directory using the `chown` command.

To view the permissions of a file or directory, you can use the `ls -l` command, which will display a long listing format that includes the permissions, ownership, and other details of each file and directory.

````python
ls -l
````

Often it is convenient to sort files by their modification time, from the oldest to the newest. This way, the newest files appear where they are most easily seen. You can do this by using the `-rt` option with the `ls` command. `t` stands for sorting by modification time, and `r` stands for reverse order (oldest first). Chaining options together is a common practice in Linux commands, allowing you to combine multiple functionalities in a single command. Typical listing command is `ls -lrt`.

````python
ls -lrt
````

# File manipulation

To copy a file, you can use the `cp` command followed by the source file and the destination file. For example, `cp file1 file2` will create a copy of `file1` named `file2`. Note that the file names can include the path to the file.

````python
cp setupQG.sh my_directory/setupQG.sh
````

You can also specify a directory as the second argument to copy the file into that directory.

````python
cp setupQG.sh my_directory
````

Note that the file `my_directory/setupQG.sh` is overwritten (without warning!) if it already exists.

To copy a directory and its contents, you can use the `-r` option with the `cp` command, which stands for "recursive". 

````python
cp -r my_directory my_directory_backup
````

A file or directory can also be just moved from one directory to another using the `mv` command.

````python
mv my_directory_backup my_directory
````

# Wildcards

Wildcards are special characters that can be used to represent one or more characters in a file name or directory name. They are commonly used in commands to specify multiple files or directories at once. The most common wildcard is the asterisk `*`, which represents any number of characters. For example, `*.py` will match all files that have a `.py` extension in the current directory.

````python
ls *.py
````

The wildcard can be placed anywhere in the file name. For example, `my_directory/*.sh` will match all files with a `.sh` extension that are located in the `my_directory` directory.

````python
ls my_directory/*.sh
````

A second common wildcard is the question mark `?`, which represents a single character. For example, `file?.txt` will match files like `file1.txt`, `fileA.txt`, but not `file10.txt`.

# Input/Output redirection

Input/output redirection is a powerful feature in Linux that allows you to redirect the input and output of commands to and from files. This can be useful for saving the output of a command to a file, or for providing input to a command from a file.

The `>` operator is used to redirect the output of a command to a file. For example, `ls > file_list.txt` will save the output of the `ls` command (which lists the contents of the current directory) to a file called `file_list.txt`. If the file already exists, it will be overwritten.

````python
ls > file_list.txt
````

If you want to append the output to an existing file instead of overwriting it, you can use the `>>` operator. For example, `ls >> file_list.txt` will add the output of the `ls` command to the end of the `file_list.txt` file.

````python
ls >> file_list.txt
````

An output of one command can be sent as input to another command using the `|` operator, which is called a pipe. For example, `ls | grep "dir"` will list the contents of the current directory and then filter the results to show only those that contain "dir".

````python
ls | grep "dir"
````

# Keyboard shortcuts in the terminal

- `Ctrl + C`: Interrupt the current command or process.

- `Ctrl + A`: Move the cursor to the beginning of the line.

- `Ctrl + E`: Move the cursor to the end of the line.

- `up arrow`: Cycle through the command history to recall previously entered commands.

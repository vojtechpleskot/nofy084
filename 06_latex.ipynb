# Remark: TeX and LaTeX

- TeX is the typesetting system created by Donald Knuth in the late 1970s. It is a low-level language that provides a lot of control over the formatting of documents.

- LaTeX is a high-level macro package built on top of TeX, created by Leslie Lamport in the 1980s. It provides a more user-friendly interface for creating documents, with predefined commands and environments for common tasks.

- TeX is like the engine of a car, while LaTeX is like the body and interior.

- Using TeX directly is possible but difficult and there is no need to do it.

# Packages

- LaTeX packages are extensions that provide additional functionality. You can use packages to add new commands, environments, and formatting options to your document.

- To use a package, you need to include it in the preamble of your LaTeX document using the `\usepackage{}` command.

- For example, to use the `hyperref` package, which allows you to create hyperlinks in your document, you would add the following line to the preamble:

    ```latex
    \usepackage{hyperref}
    ```

# Bibliography

- LaTeX has a powerful bibliography management system called BibLaTeX. It allows you to create a bibliography database and cite references in your document.

- Nice intro to BibLaTeX: [https://cs.overleaf.com/learn/latex/Bibliography_management_with_biblatex](https://cs.overleaf.com/learn/latex/Bibliography_management_with_biblatex)

- Five main steps to use BibLaTeX:

    1. Use the `biblatex` package in your LaTeX document by adding a command like the following in the preamble:

        ```latex
        \usepackage[
          backend=biber,
          style=numeric,
          sorting=nyt,
          giveninits=true
        ]{biblatex}
        ```
    
    2. Create a bibliography database file (e.g., `references.bib`) and add your references in BibLaTeX format.

    3. Include the bibliography in your LaTeX document using the `\addbibresource{references.bib}` command in the preamble.

    4. Cite references in your document using the `\cite{key}` command, where `key` is the identifier of the reference in your bibliography file.

    5. Print the bibliography before the Appendix (if any) using the `\printbibliography` command.

- A nice BibLaTeX entry looks like this:

    ```bibtex
    @Article{higgs-discovery-atlas,
        author         = "{ATLAS Collaboration}",
        title          = "{Observation of a new particle in the search for the Standard Model Higgs boson with the ATLAS detector at the LHC}",
        journal        = "Phys. Lett. B",
        volume         = "716",
        year           = "2012",
        pages          = "1",
        doi            = "10.1016/j.physletb.2012.08.020",
        reportNumber   = "CERN-PH-EP-2012-218",
        eprint         = "1207.7214",
        archivePrefix  = "arXiv",
        primaryClass   = "hep-ex",
    }
    ```

    - Not all the fields are required in each use-case, however.

    - In the text, you can cite this reference using `\cite{higgs-discovery-atlas}`.

- You can use online tools like [Zotero](https://www.zotero.org/) to manage your bibliography and export it in BibLaTeX format. Very often, it is enough to just click a "download citation" link at the desired scientific article web page and copy-paste the BibLaTeX or BibTeX (both are in the same format) entry into your bibliography `.bib` file.

# Including content from other .tex files

- Use the `\input{}` command. This helps you organize your document into multiple files.

- For example, if you have a file called `introduction.tex` that contains the introduction section of your document, you can include it in your main LaTeX file using the following command:

    ```latex
    \input{introduction.tex}
    ```

    - IMPORTANT: If `introduction.tex` was in a subfolder, e.g. called `source`, then we would write `\input{source/introduction.tex}`.

- You can keep e.g. tables in separate files and include them in the appropriate place. This is particularly useful if you generate tables automatically from data using a script. You can have the script generate a `.tex` file with the table and then include it in any `.tex` file using `\input{}`.

#### Strings in Python, brief summary

- f-strings are essential for including data (stored in variables) in strings.

    - Include a variable value in a string using the braces `{}` syntax. For example:

````python
name, age = "Alice", 30
row = f"{{{name}}} & {age} \\\\"
print(row)
````

- To include a literal backslash `\` in a string, you need to escape it with another backslash (i.e., `\\`).

    - You can also use raw strings (e.g., `r"Path\to\file"`) to avoid having to escape backslashes.

- To include a literal curly brace `{` or `}` in an f-string, you need to double it (i.e., `{{` or `}}`).

- f-strings allow for powerful rounding!

````python
value = 3.141592653589793
print(f"Value rounded to 2 decimal places: {value:.2f}")
print(f"Value rounded to 2 significant figures: {value:.2g}")
````

- Writing a LaTeX table to a `.tex` file:

````python
name, age = "Alice", 30
my_table = \
f"""
\\begin{{tabular}}{{|c|c|}}
\\hline
Name & Age \\\\
\\hline
{name} & {age} \\\\
\\hline
\\end{{tabular}}
"""
with open('table_1.tex', 'w', encoding = "utf-8") as f:
    f.write(my_table)
````
